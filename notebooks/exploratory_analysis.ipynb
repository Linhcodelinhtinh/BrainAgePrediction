# Exploratory Data Analysis (EDA) - Brain Age Prediction

Notebook này phân tích dữ liệu thuộc tính của não bộ dùng để dự đoán tuổi não.

```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Load dữ liệu mẫu
data_path = os.path.join("..", "data", "sample_data", "brain_data.csv")
df = pd.read_csv(data_path)
df.head()
```

```python
# Xem thông tin tổng quan của dữ liệu
df.info()
```

```python
# Thống kê mô tả
df.describe()
```

```python
# Phân phối của thuộc tính 'age' (Biến mục tiêu)
plt.figure(figsize=(8, 5))
sns.histplot(df['age'], kde=True, bins=20, color='skyblue')
plt.title('Phân Phối Tuổi Não (Age Distribution)')
plt.xlabel('Age')
plt.ylabel('Count')
plt.show()
```

```python
# Ma trận tương quan giữa các thuộc tính
plt.figure(figsize=(8, 6))
sns.heatmap(df.corr(), annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Ma Trận Tương Quan (Correlation Matrix)')
plt.show()
```
